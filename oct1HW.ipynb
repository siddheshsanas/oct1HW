2. Change the case

   Ask for a word and print its first and last characters.

### Medium
6. Palindrome checker
   Ask for a word and check whether it reads the same forwards and backwards. Treat uppercase and lowercase letters as the same.
   Examples: Level → palindrome; Python → not a palindrome
7. Count vowels and consonants
   Ask for a sentence and count its vowels and consonants. Ignore spaces, digits, and punctuation.
8. Replace spaces
   Ask for a sentence and replace each space with a hyphen (-).
   Example: I like Python → I-like-Python
9. Find the longest word
   Ask for a sentence and print the longest word. If multiple words share the longest length, print the first one.
10. Word frequency
    Ask for a sentence and a word, then count how many times that word occurs, ignoring uppercase/lowercase differences.

```python
#6
s  = "facebook" 
a = "level"
if s == s[::-1]:
    print("pallindrome")
elif s != s[::-1]:
    print("not a pallindrome")
if a == a[::-1]:
    print("pallindrome")
elif a != a[::-1]:
    print("not a pallindrome")
```

```python
#7
sentence = 'python programming language'

vowels = "aeiou"
vowel_count = 0
consonant_count = 0

for character in sentence.lower():
    if character in vowels:
        vowel_count += 1
    elif character.isalpha():
        consonant_count += 1

print("Vowels:", vowel_count)
print("Consonants:", consonant_count)
```

```python
#9
sentence = ("python is intresting and easy to learn")
words = sentence.split()

longest_word = ""

for word in words:
    clean_word = word.strip(".,!?;:'\"()[]{}")  # Remove punctuation from the word

    if len(clean_word) > len(longest_word):
        longest_word = clean_word

if longest_word:
    print("Longest word:", longest_word)
else:
    print("No words found.")
```

```python
#10
sentence = ("instagram and instagram is the social media and instagram is the social media")
target_word = input("Enter the word to count: ").lower()
count = 0
punctuation = ".,!?;:'\"()[]{}"

for word in sentence.split():
    clean_word = word.strip(punctuation).lower()

    if clean_word == target_word:
        count += 1

print("Count:", count)
```
